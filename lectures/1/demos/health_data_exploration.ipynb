# Exploratory Analysis of Patient Data

This notebook demonstrates how to analyze and visualize health data using Python.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Set style for plots
plt.style.use('seaborn')
```

## Loading Patient Data

We'll create a sample dataset of patient information.

```python
# Create sample data
data = {
    'patient_id': range(1, 101),
    'age': np.random.normal(65, 12, 100).astype(int),
    'systolic_bp': np.random.normal(130, 15, 100).astype(int),
    'diastolic_bp': np.random.normal(80, 10, 100).astype(int),
    'cholesterol': np.random.normal(210, 30, 100).astype(int),
    'glucose': np.random.normal(100, 25, 100).astype(int),
    'smoker': np.random.choice([0, 1], 100, p=[0.7, 0.3])
}

# Create DataFrame
df = pd.DataFrame(data)

# Display first few rows
df.head()
```

## Data Summary and Exploration

```python
# Summary statistics
df.describe()
```

```python
# Create derived features
df['pulse_pressure'] = df['systolic_bp'] - df['diastolic_bp']
df['hypertension'] = (df['systolic_bp'] >= 130) | (df['diastolic_bp'] >= 80)
df['high_cholesterol'] = df['cholesterol'] > 200

# Count patients with conditions
print(f"Patients with hypertension: {df['hypertension'].sum()} ({df['hypertension'].mean()*100:.1f}%)")
print(f"Patients with high cholesterol: {df['high_cholesterol'].sum()} ({df['high_cholesterol'].mean()*100:.1f}%)")
print(f"Smokers: {df['smoker'].sum()} ({df['smoker'].mean()*100:.1f}%)")
```

## Data Visualization

```python
# Age distribution
plt.figure(figsize=(10, 6))
plt.hist(df['age'], bins=20, color='skyblue', edgecolor='black')
plt.title('Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.grid(alpha=0.3)
plt.show()
```

```python
# Blood pressure scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df['systolic_bp'], df['diastolic_bp'], 
            c=df['age'], cmap='viridis', alpha=0.7)
plt.colorbar(label='Age')
plt.title('Blood Pressure Measurements')
plt.xlabel('Systolic BP (mmHg)')
plt.ylabel('Diastolic BP (mmHg)')

# Add reference lines for hypertension
plt.axvline(x=130, color='red', linestyle='--', alpha=0.5)
plt.axhline(y=80, color='red', linestyle='--', alpha=0.5)
plt.grid(alpha=0.3)
plt.show()
```

## Relationship Between Variables

```python
# Correlation matrix
correlation = df[['age', 'systolic_bp', 'diastolic_bp', 'cholesterol', 'glucose', 'pulse_pressure']].corr()

# Plot correlation matrix
plt.figure(figsize=(10, 8))
plt.imshow(correlation, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label='Correlation')
plt.title('Correlation Between Health Metrics')

# Add labels
labels = ['Age', 'Systolic BP', 'Diastolic BP', 'Cholesterol', 'Glucose', 'Pulse Pressure']
plt.xticks(range(len(labels)), labels, rotation=45)
plt.yticks(range(len(labels)), labels)

# Add correlation values
for i in range(len(labels)):
    for j in range(len(labels)):
        plt.text(j, i, f'{correlation.iloc[i, j]:.2f}', 
                 ha='center', va='center', color='white' if abs(correlation.iloc[i, j]) > 0.5 else 'black')

plt.tight_layout()
plt.show()
```
